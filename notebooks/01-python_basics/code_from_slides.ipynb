# 5. Lists and indexing

## Lists hold collections of texts: example

```python
texts = [
    "We call for dialogue.",
    "We urge an immediate ceasefire."
]
print(texts)
print(len(texts))
```

## Zero-based and negative indexing: example

```python
topics = ["dialogue", "ceasefire", "humanitarian aid"]
print(topics[0])   # dialogue
print(topics[1])   # ceasefire
print(topics[-1])  # humanitarian aid
```

## Slices select several items: example

```python
topics = ["dialogue", "ceasefire", "humanitarian aid"]
print(topics[:2])   # ['dialogue', 'ceasefire']
print(topics[1:3])  # ['ceasefire', 'humanitarian aid']
print(topics[1:])   # ['ceasefire', 'humanitarian aid']
```

## Appending items: example

```python
labels = ["dialogue"]
labels.append("ceasefire")
print(labels)
# ['dialogue', 'ceasefire']
```

## Splitting text into a list: example

```python
text = "  We call for dialogue.  "
clean_text = text.strip()
words = clean_text.split()
print(words)
# ['We', 'call', 'for', 'dialogue.']
```

## Joining text from a list: example

```python
parts = ["We", "call", "for", "dialogue."]
text = " ".join(parts)
prompt = "Read this statement: " + text
print(text)
print(prompt)
```

# 8. Conditions and loops

## Conditions and indentation: example

```python
score = 0.85
if score > 0.8:
    print("Inspect this high-scoring text.")
else:
    print("Inspect this lower-scoring text.")
```

## Membership checks: example

```python
text = "We call for dialogue."
allowed_labels = ["dialogue", "ceasefire"]
settings = {"batch_size": 16}
print("dialogue" in text)           # True
print("dialogue" in allowed_labels) # True
print("batch_size" in settings)    # True
```

## For loops repeat an operation: example

```python
texts = ["  Peace talks.  ", "  A ceasefire now.  "]
for text in texts:
    print(text.strip())
# Peace talks.
# A ceasefire now.
```

## Indentation: a working example

```python
texts = [
    "We support peaceful dialogue.",
    "A ceasefire now.", 
]
for text in texts:
    print(text)
    print("\tThis line belongs to the loop.")
print("This line is outside the block.")
```

## Collecting results in a loop: example

```python
texts = ["Peace talks.", "A ceasefire now."]
lengths = []
for text in texts:
    lengths.append(len(text))
print(lengths)
# [12, 15]
```
